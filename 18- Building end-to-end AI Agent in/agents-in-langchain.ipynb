```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import requests
from dotenv import load_dotenv

load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.2,
)
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search = DuckDuckGoSearchRun()

print(search.invoke("What is current news of india on hindu?"))
```

```python
from langchain.agents import create_agent
```

```python
react_agent = create_agent(
    model=model,
    tools=[search],
    system_prompt="""You are a helpful assistant. Use the available tools when necessary to answer the user's question."""
    # system_prompt=prompt.template
)
```

```python
response = react_agent.invoke({
    "messages": [
        ("user", "Current news of india on hindu?")
    ]
})
```
