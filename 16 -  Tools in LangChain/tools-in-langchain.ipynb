## Custom Tools


```python
from langchain_core.tools import tool
```

```python
def Addition(a,b):
    '''Addition of two numbers'''
    return a+b
```

```python
# add type hints to the function

def Addition(a :int,b :int) -> int:
    '''Addition of two numbers'''
    return a+b
```

```python
# Final Tool

@tool
def Addition(a :int,b :int) -> int:
    '''Addition of two numbers'''
    return a+b
```

```python
print(Addition.invoke({'a': 5, 'b': 10}))
```

```python
# these are three attributes of the tool function
print(Addition.name)
print(Addition.description)
print(Addition.args)
```

```python
print(Addition.args_schema.model_json_schema())

# This is what LLM gets
```

## Method 2 - Using StructuredTool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class AdditionInput(BaseModel):
    a: int = Field(required=True, description="The first number to Add")
    b: int = Field(required= True, description="The second number to Add")
```

```python
def Addition_func(a: int, b: int) -> int:
    return a + b
```

```python
Addition_tool = StructuredTool.from_function(
    func = Addition_func,
    name  = "Addition",
    description = "This tool adds two numbers",
    args_schema = AdditionInput
)
```

```python
result = Addition_tool.invoke({'a':3, 'b':6})
print(result)
print(Addition_tool.name)
print(Addition_tool.description)
print(Addition_tool.args)
```

## Method 3 - Using BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':6, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

## Toolkit

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
